```python
# 1. raw dirty database
raw_names = ["  iysa ", "  ALICE", "bob   ", "   diana  "]

# 2. Function decloration for cleaning a single string
def format_name(name):
    cleaned = name.strip().capitalize()
    return cleaned

# 3. Empty list to store processed names
clean_names = []

# 4. Iterating over raw names and applying the function
for raw_name in raw_names:
    formatted = format_name (raw_name) # Pass dirty name into the function
    clean_names.append(formatted) # append clean result to the list

# 5ю Output the cleaned dataset
print("Original names:", raw_names)
print("clean_names:", clean_names)
```
