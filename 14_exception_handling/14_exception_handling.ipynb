```python
# ===================================================
# Module 14: Robust Error Handling & Custom Exceptions
# ===================================================

# 1. Defining a custom user-defined exception class
class InvalidScoreError(Exception):
    """Custom exception raised when a score falls outside allowable bounds (0-100)."""
    pass


# 2. Data processing function with comprehensive exception blocks
def parse_student_record(raw_record):
    """Parses raw string record (Name,Score) with robust exception guards."""
    try:
        name, score_str = raw_record.split(",")
        score = float(score_str)

        # Explicit validation check triggering custom exception
        if score < 0 or score > 100:
            raise InvalidScoreError(f"Score {score} is out of range (0-100).")

    except ValueError as e:
        # Handles wrong string-to-float conversions or missing commas
        print(f"[ERROR] Type Conversion Failed for '{raw_record}': {e}")
        return None

    except InvalidScoreError as e:
        # Handles custom domain logic violations
        print(f"[ERROR] Range Validation Failed for '{raw_record}': {e}")
        return None

    else:
        # Executes ONLY if no exception occurred in the try block
        return {"Name": name.strip(), "Score": score}

    finally:
        # Executes ALWAYS regardless of success or exception state
        pass


# --- 3. Executing Test Cases over Unclean Dataset ---
unclean_dataset = [
    "Iysa, 95",
    "Alice, eighty-five",  # Triggers ValueError
    "Bob, 150",            # Triggers InvalidScoreError
    "Diana, 88"
]

print("--- 1. Parsing Raw Unclean Dataset ---")
cleaned_records = []

for raw_item in unclean_dataset:
    result = parse_student_record(raw_item)
    if result:
        cleaned_records.append(result)

print("\n--- 2. Final Cleaned Records Output ---")
print(cleaned_records)
```
