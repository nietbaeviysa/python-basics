```python
# ===================================================
# Module 11: Memory-Efficient Data Processing via Generators
# ===================================================

# Simulated large dataset of raw student log entries
raw_logs = [
    "Iysa,95,Passed",
    "Alice,82,Passed",
    "Bob,45,Failed",
    "Diana,91,Passed",
    "Ethan,68,Failed"
]


# 1. Generator function using yield for lazy evaluation
def process_logs_generator(logs):
    """Processes log records one by one without loading all data into memory at once."""
    for record in logs:
        name, score, status = record.split(",")
        # Yield returns data control to caller and pauses function execution state
        yield {
            "Name": name,
            "Score": int(score),
            "Status": status
        }


# --- 2. Consuming Generator via Iteration ---
print("--- 1. Iterating Over Generator Stream ---")
# Creating the generator object (no calculations performed yet)
log_stream = process_logs_generator(raw_logs)

print("Generator Object Created:", log_stream)

# Processing elements on-the-fly inside loop
for entry in log_stream:
    if entry["Score"] >= 80:
        print(f"High Achiever: {entry['Name']:<6} | Score: {entry['Score']}")


# --- 3. Consuming Generator with next() ---
print("\n--- 2. Fetching Single Items manually using next() ---")
# Re-instantiating fresh generator
fresh_stream = process_logs_generator(raw_logs)

first_student = next(fresh_stream)
second_student = next(fresh_stream)

print("First Item Fetched: ", first_student)
print("Second Item Fetched:", second_student)
```
