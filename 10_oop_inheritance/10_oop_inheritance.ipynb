```python
# ===================================================
# Module 10: OOP Inheritance & Method Overriding
# ===================================================

# 1. Base Class (Parent)
class Student:
    """Base class representing a standard student record."""
    def __init__(self, name, city):
        self.name = name
        self.city = city
        self.scores = []

    def add_score(self, score):
        """Appends a new score to the record."""
        self.scores.append(score)

    def get_average(self):
        """Calculates the mean score."""
        return sum(self.scores) / len(self.scores) if self.scores else 0.0

    def get_status(self):
        """Determines pass status based on standard criteria."""
        return "Passed" if self.get_average() >= 70 else "Failed"


# 2. Derived Class (Child) inheriting from Student
class GraduateStudent(Student):
    """Child class extending Student with research topic and stricter status rules."""

    def __init__(self, name, city, research_topic):
        # Call parent constructor to initialize inherited attributes
        super().__init__(name, city)
        self.research_topic = research_topic

    # Method Overriding: Redefining get_status with stricter academic standards
    def get_status(self):
        avg = self.get_average()
        if avg >= 85:
            return "Graduated with Honors"
        elif avg >= 75:
            return "Passed"
        return "Needs Thesis Revision"


# --- 3. Instantiation and Execution ---
print("--- 1. Parent Class Execution ---")
student = Student("Alice", "Samarkand")
student.add_score(75)
student.add_score(80)
print(f"Student: {student.name} | Status: {student.get_status()}")


print("\n--- 2. Derived Class (Inheritance) Execution ---")
grad_student = GraduateStudent("Iysa", "Tashkent", "Data Engineering Pipelines")
grad_student.add_score(90)
grad_student.add_score(92)

print(f"Graduate Student: {grad_student.name}")
print(f"Research Topic:   {grad_student.research_topic}")
print(f"Average Score:    {grad_student.get_average():.1f}")
print(f"Academic Status:  {grad_student.get_status()}")
```
