```python
# ===================================================
# Module 15: Python Modules, Namespaces & Script Entry Points
# ===================================================

import math
from datetime import datetime


# 1. Utility class simulating an external module tool
class AnalyticsEngine:
    """Provides statistical calculations and metadata tracking."""

    @staticmethod
    def calculate_std_dev(scores):
        """Calculates population standard deviation using math module."""
        if not scores or len(scores) < 2:
            return 0.0

        mean = sum(scores) / len(scores)
        variance = sum((x - mean) ** 2 for x in scores) / len(scores)
        return math.sqrt(variance)

    @staticmethod
    def generate_report_header(title):
        """Generates a timestamped header string using datetime module."""
        timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        return f"=== {title.upper()} | Generated at: {timestamp} ==="


# 2. Main execution function guarded by script entry logic
def main():
    print("--- 1. Testing Imported Module Functions ---")

    student_scores = [85, 92, 78, 95, 88]

    # Utilizing module methods
    header = AnalyticsEngine.generate_report_header("Student Performance Analysis")
    std_dev = AnalyticsEngine.calculate_std_dev(student_scores)

    print(header)
    print(f"Dataset: {student_scores}")
    print(f"Mean Score: {sum(student_scores)/len(student_scores):.2f}")
    print(f"Standard Deviation: {std_dev:.2f}")


# 3. Standard Python entry point guard idiom (__name__ == '__main__')
if __name__ == "__main__":
    main()
```
