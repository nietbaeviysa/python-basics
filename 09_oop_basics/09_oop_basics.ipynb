```python
# ===================================================
# Module 09: Object-Oriented Programming (OOP) Basics
# ===================================================

# 1. Defining a Student class blueprint
class Student:
    """Represents a student record with dynamic score management."""

    # Constructor method to initialize object attributes
    def __init__(self, name, city):
        self.name = name
        self.city = city
        self.scores = []  # Encapsulated list of numerical scores

    # Method to append new test scores
    def add_score(self, score):
        """Appends a new score to the student's record."""
        if 0 <= score <= 100:
            self.scores.append(score)
        else:
            print(f"Invalid score ({score}) for {self.name}. Must be between 0 and 100.")

    # Method to calculate average performance
    def get_average(self):
        """Calculates and returns the mean score."""
        if not self.scores:
            return 0.0
        return sum(self.scores) / len(self.scores)

    # Dunder/Magic method for clean human-readable output
    def __str__(self):
        avg = self.get_average()
        return f"Student: {self.name:<7} | City: {self.city:<9} | Avg Score: {avg:.1f}"


# --- 2. Object Instantiation and Usage ---
print("--- 1. Creating Student Objects ---")
# Creating instances of the Student class
student1 = Student("Iysa", "Tashkent")
student2 = Student("Alice", "Samarkand")

# Adding scores via object methods
student1.add_score(95)
student1.add_score(88)
student1.add_score(92)

student2.add_score(78)
student2.add_score(82)

# Displaying formatted representation (__str__)
print(student1)
print(student2)


# --- 3. Evaluating Performance Metrics ---
print("\n--- 2. Academic Performance Check ---")
for student in [student1, student2]:
    status = "Honor Roll" if student.get_average() >= 90 else "Standard Pass"
    print(f"{student.name} Academic Status: {status}")
```
