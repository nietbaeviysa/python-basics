```python
student_scores = {
    "Alice": 85,
    "Bob": 92,
    "Charlie": 78,
    "Diana": 95,
    "Ethan": 80
}
passed_students = []
for student,score in student_scores.items():
  if score  >= 85:
    passed_students.append(student)


print("students who passed")
for student in passed_students:
  print(student)
```

```python
# 1. Dictionary initialization: storing student names and their test scores
student_scores = {
    "Alice": 85,
    "Bob": 92,
    "Charlie": 78,
    "Diana": 95,
    "Ethan": 80
}

# 2. Iterating through dictionary values to display individual scores
print("--- Student List ---")
for score in student_scores.values():
    print("score:", score)

# 3. Data Aggregation: calculating total sum, total count, and average score
total_sum = sum(student_scores.values())  # Calculate the sum of all scores
count = len(student_scores)               # Get the total number of students
average = total_sum / count               # Compute the average score


print("\nAverage score of group:", average)
```
