```python
# Advanced Functions: Default Arguments & Lambdas
# ===================================================

# 1. Function declaration with a default argument (bonus=5)
def calculate_final_score(score, bonus=5):
    """Adds a bonus value to the base score (default bonus is 5)."""
    final_score = score + bonus
    # Prevent score from exceeding the 100 limit
    if final_score > 100:
        return 100
    return final_score


# --- Test 1: Testing Default vs Custom Arguments ---
print("--- 1. Default vs Custom Arguments ---")
# Uses default bonus=5
print("Score 75 with default bonus:", calculate_final_score(75))

# Overrides default bonus with bonus=10
print("Score 75 with custom bonus (10):", calculate_final_score(75, bonus=10))


# --- Test 2: Lambda function with filter() ---
raw_scores = [70, 85, 40, 90, 78, 95]

# Extracting only scores that are strictly less than 80
low_scores = list(filter(lambda s: s < 80, raw_scores))

print("\n--- 2. Filtering Low Scores (< 80) ---")
print("Raw dataset:", raw_scores)
print("Low scores:", low_scores)


# --- Test 3: Lambda function with map() ---
# Applying calculate_final_score with default bonus to every item in the list
processed_scores = list(map(lambda s: calculate_final_score(s), raw_scores))

print("\n--- 3. Bulk Processing with Map & Lambda ---")
print("Scores after applying bonus:", processed_scores)
```
