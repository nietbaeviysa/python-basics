```python
# ===================================================
# Module 13: JSON Data Handling & Nested Structure Parsing
# ===================================================

import json

# 1. Simulated raw JSON payload (string format) received from an API response
json_payload = """
{
    "status": 200,
    "data": {
        "course": "Python for Data Science",
        "students": [
            {"id": 1, "name": "Iysa", "scores": [90, 95, 92], "active": true},
            {"id": 2, "name": "Alice", "scores": [85, 88, 80], "active": true},
            {"id": 3, "name": "Bob", "scores": [60, 65, 58], "active": false}
        ]
    }
}
"""

# 2. Parsing JSON string into a native Python dictionary using json.loads()
parsed_data = json.loads(json_payload)

# --- 3. Extracting and Parsing Nested Data Structures ---
print("--- 1. Extracted API Metadata ---")
print("Response Status:", parsed_data["status"])
print("Course Title:   ", parsed_data["data"]["course"])

print("\n--- 2. Active Student Performance Analytics ---")
active_summary = []

# Iterating through nested list of dictionaries
for student in parsed_data["data"]["students"]:
    if student["active"]:
        # Calculate dynamic score average from nested array
        scores = student["scores"]
        avg_score = sum(scores) / len(scores)

        active_summary.append({
            "Name": student["name"],
            "AverageScore": round(avg_score, 1)
        })

for item in active_summary:
    print(f"Student: {item['Name']:<6} | Mean Score: {item['AverageScore']}")


# --- 4. Serializing Python Dict back to Formatted JSON (json.dumps) ---
exported_json = json.dumps(active_summary, indent=4)

print("\n--- 3. Formatted Exported JSON Output ---")
print(exported_json)
```
