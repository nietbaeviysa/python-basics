```python
# ===================================================
# Module 12: Python Decorators (@decorator)
# ===================================================

import time
import functools

# 1. Defining a custom decorator function
def execution_logger(func):
    """Decorator that logs execution start/finish and measures processing time."""
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.time()
        print(f"[LOG] Executing function: '{func.__name__}'...")

        # Execute target function
        result = func(*args, **kwargs)

        duration = (time.time() - start_time) * 1000  # Convert to milliseconds
        print(f"[LOG] Finished '{func.__name__}' in {duration:.2f} ms\n")
        return result

    return wrapper


# 2. Applying decorator to a data processing function using @ syntax
@execution_logger
def process_student_batch(students):
    """Simulates processing a batch of student records with workload delay."""
    results = []
    for name, score in students:
        time.sleep(0.05)  # Simulate I/O or network delay (50ms per item)
        status = "Honor Roll" if score >= 90 else "Passed"
        results.append({"Name": name, "Score": score, "Status": status})
    return results


# --- 3. Execution ---
dataset = [
    ("Iysa", 95),
    ("Alice", 88),
    ("Bob", 92),
    ("Diana", 78)
]

print("--- Running Decorated Function ---")
batch_output = process_student_batch(dataset)

print("--- Execution Results ---")
for record in batch_output:
    print(record)
```
