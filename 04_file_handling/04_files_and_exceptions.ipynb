```python
clean_names = ["Iysa", "Alice", "Bob", "Diana"]

with open("student.txt", "w") as file:
  for name in clean_names:
    file.write(name + "\n")

print("Data successfully saved to student.txt")
```

```python
# Исходные данные
student_logs = ["Alice: 90", "Bob: 85", "Charlie: 78", "Diana: 92"]
file_path = "journal.txt"

# 1. Запись в файл
with open(file_path, "w") as file:
    for name in student_logs:
        file.write(name + "\n")

# 2. Чтение из файла и обработка ошибок
try:
    with open(file_path, "r") as file:
        read_logs = file.read().splitlines()

    print("Прочитанные записи:", read_logs)
    print("Всего записей в журнале:", len(read_logs))

except FileNotFoundError:
    print(f"Ошибка: Файл '{file_path}' не найден!")
```
