```python

# Module 08: List & Dictionary Comprehensions

# 1. Raw dataset: List of student records (tuples with Name, Score)
students_data = [
    ("Iysa", 85),
    ("Alice", 92),
    ("Bob", 65),
    ("Diana", 78),
    ("Ethan", 88)
]

# 2. List Comprehension: Filter students with score >= 80 and convert names to UPPERCASE
top_students = [name.upper() for name, score in students_data if score >= 80]

# 3. Dictionary Comprehension: Transform tuples into a dict with Pass/Fail status
# Assign 'Pass' if score >= 70, else 'Fail'
status_dict = {name: ("Pass" if score >= 70 else "Fail") for name, score in students_data}

# --- Output Execution Results ---
print("--- 1. List Comprehension Result ---")
print("Top Students (Score >= 80):", top_students)

print("\n--- 2. Dictionary Comprehension Result ---")
for student, status in status_dict.items():
    print(f"Student: {student:<7} | Status: {status}")
```
