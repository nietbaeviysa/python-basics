```python
# 1. working with Lists and the .append () method

# Initializing a list of academic subjects
subjects = ["Data Science", "Python Basics","Statistics"]

# Adding a new element to the end of the list
subjects.append("Machine Learning")

# Iterating throught the list using a for loop
print("---My Learning Subjects---")
for subject in subjects:
  print(" Currently studying :", subject)

#====================================================
# 2 Data Filtering (for loop + if + .append)

# Initial list of values
scores = [ 5,12,8,9,15,17,19,27]

# Emty list to scorefiltered results
high_scores = []

# Loop through each score and filter values greater than 10
for score in scores:
  if score > 10:
    high_scores.append(score)

print("\n--- Filtering Scores ---")
print("Original score:", scores)
print("Scores greater than 10:", high_scores)
```
