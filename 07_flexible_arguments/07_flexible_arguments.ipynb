```python
# Module 07: Flexible Function Arguments (*args, **kwargs)
# ===================================================

# 1. Function using *args to aggregate variable numerical inputs
def calculate_average_score(*scores):
    """Calculates the average score from any number of positional arguments."""
    if not scores:
        return 0
    total = sum(scores)
    count = len(scores)
    return total / count


# 2. Function using **kwargs to process dynamic keyword metadata
def build_student_profile(name, **metadata):
    """Creates a student profile dictionary with dynamic key-value pairs."""
    profile = {"Name": name}

    # Iterating over flexible keyword arguments
    for key, value in metadata.items():
        profile[key] = value

    return profile


# --- Test 1: Testing *args with different numbers of inputs ---
print("--- 1. Aggregating Scores with *args ---")
# Passing 3 scores
avg_3 = calculate_average_score(80, 90, 85)
print("Average of 3 scores:", round(avg_3, 2))

# Passing 5 scores to the same function
avg_5 = calculate_average_score(70, 88, 92, 60, 95)
print("Average of 5 scores:", round(avg_5, 2))


# --- Test 2: Testing **kwargs with variable metadata ---
print("\n--- 2. Building Student Profiles with **kwargs ---")
# Passing name + 3 custom key-value pairs
student_1 = build_student_profile("Iysa", age=19, city="Tashkent", status="Active")

# Passing name + 2 different custom key-value pairs
student_2 = build_student_profile("Alice", city="Samarkand", grade="A")

print("Student 1 Profile:", student_1)
print("Student 2 Profile:", student_2)
```
